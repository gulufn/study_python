```python
# import tensorflow as tf
import numpy as np
import math
import pandas as pd
import matplotlib.pyplot as plt
%matplotlib inline
```

```python
from PIL import Image
```

```python
img = Image.open('./asses/img/a.jpg').convert('RGBA')
w, h = img.size
px = img.load()

# 图A：保留偶数列，奇数列透明
a = img.copy()
pa = a.load()
for x in range(w):
    if x % 2 == 1:
        for y in range(h):
            r, g, b, _ = pa[x, y]
            pa[x, y] = (r, g, b, 0)
a.save('./makeshift/a.png')
```

```python
# 图B：保留奇数列，偶数列透明
b = Image.open('./asses/img/b.jpg').convert('RGBA')
pb = b.load()
for x in range(w):
    if x % 2 == 0:
        for y in range(h):
            r, g, b_, _ = pb[x, y]
            pb[x, y] = (r, g, b_, 0)
b.save('./makeshift/b.png')
```
